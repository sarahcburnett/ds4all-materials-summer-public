## 2. Defining Functions

So far you've *called* functions like `max()`, `abs()`, and `print()`. Now let's *define* one. Creating a function is just like creating any other R object: you give it a name and assign it a value with `<-`. The value happens to be a function.

Here is the world's simplest useful function. Run the cell:

````r
double <- function(x) {
  2 * x
}

double(5)
````

Every function definition has the same anatomy:

- `function(x)` says this function takes one **parameter** named `x`;
- the **body** between `{` and `}` is the code that runs when the function is called;
- the value of the **last evaluated expression** in the body is what the function **returns**. <br> You can also write `return(2 * x)` explicitly, but R programmers usually rely on the last-expression rule.

When you call `double(5)`, R creates a temporary environment where `x` is `5`, runs the body, and hands back the result. Then `x` disappears; it was **local** to the function (remember scope from the [Functions](https://dats.columbian.gwu.edu/ds4all/modules/functions.html) module).

### 2.2. Printing is not returning

Run the next cell and look carefully at what happens.

````r
print_percentage <- function(proportion) {
  print(proportion * 100)
}

result <- print_percentage(0.2)
result
````

You should see `20` printed *twice*, but for two different reasons! The first `20` was printed by `print()` while the function ran. The second appeared because... `print()` also *returns* its argument, so `result` got the value 20. That's a quirk of `print()`. Many functions that print things return nothing useful at all.

The distinction matters: **returned** values can be stored in variables and used in later computations. **Printed** output just appears on the screen and is gone. When you write functions, return your answer; let the caller decide whether to print it. (In the [Functions 2](https://dats.columbian.gwu.edu/ds4all/modules/functional_programming.html) module we called functions like this *pure functions*.)

## 3. Control Flow: Making Decisions

Code that always does the same thing can only get you so far. **Control flow** lets your code branch: do this if a condition holds, otherwise do that.

```r
if (boolean expression) {
  ...
} else if (boolean expression) {
  ...
} else {
  ...
}
```

R checks the conditions from top to bottom and runs the **first** branch whose condition is `TRUE`, skipping the rest. Run this example from the Control Flow module:

````r
temp_f <- 85

if (temp_f >= 90) {
  "Hot"
} else if (temp_f >= 70) {
  "Warm"
} else {
  "Cool"
}
````

### 3.3. Branching on a whole vector: `ifelse()`

`if` looks at a **single** `TRUE`/`FALSE`. But data usually comes as a whole vector of values. For that, R gives us the **vectorized** conditional `ifelse(test, yes, no)`: it checks the condition for *every element* and builds a new vector of the same length.

````r
x <- c(-2, -1, 0, 1, 2)
ifelse(x >= 0, "nonnegative", "negative")
````

### 6.1. Integers are their own type

R has two ways to store the number 3: as a **double** (`3`) and as an **integer** (`3L`, the `L` makes it an integer). They usually behave the same, but `typeof()` tells them apart, and autograders check `typeof()`!

Run this cell:

````r
typeof(3)
typeof(3L)
typeof(3L + 1L)
typeof(3L + 1)    # mixing integer and double gives a double!
typeof(as.integer(3.0))
````

Now for a genuine trap. Money amounts like $19.99 look exact, but computers store doubles in binary, and `19.99 * 100` is secretly a hair *less* than 1999. `as.integer()` **truncates** (chops toward zero) rather than rounding. 

Watch:

````r
19.99 * 100                # prints as 1999...
as.integer(19.99 * 100)   # ...but truncates to 1998!
````

The fix: `round()` first, *then* convert with `as.integer()`.
